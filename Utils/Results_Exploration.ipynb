# Graphing Predicted vs. Actuals

### A function for making an interactive graph with plotly

Note that this requires the keys y_true_max and y_pred_max (rather than y_true and y_pred)

```python
def make_interactive_graph(preds_df):
    # Create a Plotly figure
    fig = go.Figure()

    # Initialize visibility list for each trace
    visible_list = []

    # Loop through each fips_code and add traces to the figure
    for fips in preds_df['fips_code'].unique():
        df_fips = preds_df[preds_df['fips_code'] == fips]

        # Convert PeriodIndex to datetime
        df_fips['datetime'] = df_fips['datetime'].dt.to_timestamp()

        # Add true values trace
        fig.add_trace(go.Scatter(
            x=df_fips['datetime'],
            y=df_fips['y_true_max'],
            mode='lines+markers',
            name=f'True Values - {fips}',
            visible=False  # Make it invisible initially
        ))

        # Add predicted values trace
        fig.add_trace(go.Scatter(
            x=df_fips['datetime'],
            y=df_fips['y_pred_max'],
            mode='lines+markers',
            name=f'Predicted Values - {fips}',
            visible=False  # Make it invisible initially
        ))

    # Add slider steps
    steps = []
    for i, fips in enumerate(preds_df['fips_code'].unique()):
        step = dict(
            method="update",
            args=[
                {"visible": [False] * len(fig.data)},  # Hide all traces first
                {"title": f"True vs Predicted for fips_code {fips}"}
            ]
        )
        
        # Set visible traces for the current fips_code
        step["args"][0]["visible"][i*2] = True  # True values for this fips_code
        step["args"][0]["visible"][i*2 + 1] = True  # Predicted values for this fips_code
        steps.append(step)

    # Add slider
    sliders = [dict(
        active=0,
        currentvalue={"prefix": "fips_code: "},
        pad={"t": 50},
        steps=steps
    )]

    # Update layout
    fig.update_layout(
        sliders=sliders,
        showlegend=True,
        title="Interactive Plot: True vs Predicted",
    )

    # Show the plot
    fig.show()
```

### Interactive graph of predictions vs actual for each county for Naive

```python
make_interactive_graph(naive_predictions)
```

### Interactive graph of predictions vs actual for each county for Linear Regression

```python
make_interactive_graph(lr_predictions)
```

### Interactive graph of predictions vs actual for each county for XGBoost

```python
make_interactive_graph(xgb_predictions)
```

Add a dictionary of FIPS codes and some variables to support graph construction

### Functions for Constructing Graphs

A function to graph residuals vs datetime, and a function to graph residuals vs FIPS code

```python
def graph_residuals_vs_datetime(df):
    # A plot of residual vs datetime
    fig = px.scatter(
        df,
        x='datetime',
        y='residual',
        title='Residuals vs. Datetime for XGBoost',
        labels={'datetime': 'Datetime (Year-Month)', 'residual': 'Residual'},
        template='plotly_white',
        opacity=0.3  # Adjust transparency to better display density
    )

    # Update x-axis to show only year and month
    fig.update_xaxes(tickformat='%Y-%m')

    fig.update_layout(height=800)

    # Show the plot
    fig.show()

def graph_residuals_vs_fips_code(df):
    # Convert fips_code to string for better display
    df['fips_code'] = df['fips_code'].astype(str)

    # Create a scatter plot
    fig = px.scatter(
        df,
        x='fips_code',
        y='residual',
        title='Residuals vs. FIPS Code',
        labels={'fips_code': 'FIPS Code', 'residual': 'Residual'},
        template='plotly_white',
        opacity=0.3  # Adjust transparency to better display density
    )
    fig.update_layout(height=800)

    # Update x-axis to only show labels for fips_code where is_first_fips is True
    tickvals = df.loc[df['is_first_fips'], 'fips_code']
    fig.update_xaxes(tickvals=tickvals)

    # Show the plot
    fig.show()
```

# Investigate results by county

### Functions for grouping by fips_code and adding additional county-level variables

```python
# Load the county-level variables
county_vars = pd.read_csv('../Data/County_level_Variables/Counties_2021.csv')

county_vars = county_vars.drop(columns=['neighbors', 'centroid_longitude', 'centroid_latitude', 'centroid_rounded'])
```

```python
# A function to g by fips_code, 
# compute the mean absolute value of residuals, 
# mean values of max percent of customers out, 
# and merge with county-level variables
def group_by_fips(df, county):
    df = df.groupby('fips_code').agg(
        mean_abs_residual=('residual', lambda x: x.abs().mean()),
        mean_pct_customers_out=('y_true_max', lambda x: x.mean())
    ).reset_index()

    df['fips_code'] = df['fips_code'].astype(int)
    county['FIPS'] = county['FIPS'].astype(int)

    merged_df = pd.merge(df, county_vars, left_on='fips_code', right_on='FIPS', how='left')

    return merged_df
```

```python
xgb_byfips = group_by_fips(xgb_predictions, county_vars)
```

### Functions for constructing graphs to compare counties

```python
# Graph mean_abs_residual vs mean_pct_customers_out
def plot_mean_abs_residual_vs_mean_pct_customers_out(df):
    fig = px.scatter(
        df,
        x='mean_pct_customers_out',
        y='mean_abs_residual',
        title='Mean Absolute Residual vs. Mean Percent of Customers Out',
        labels={'mean_pct_customers_out': 'Mean Percent of Customers Out', 'mean_abs_residual': 'Mean Absolute Residual'},
        template='plotly_white'
    )

    fig.update_layout(height=800)

    # Show the plot
    fig.show()

# Graph mean_abs_residual vs population
def plot_mean_abs_residual_vs_population(df):
    fig = px.scatter(
        df,
        x='POPULATION',
        y='mean_abs_residual',
        title='Mean Absolute Residual vs. Population',
        labels={'POPULATION': 'Population', 'mean_abs_residual': 'Mean Absolute Residual'},
        template='plotly_white'
    )

    fig.update_layout(height=800)

    # Show the plot
    fig.show()

# Graph mean_abs_residual vs percent of buried power lines
def plot_mean_abs_residual_vs_pct_buried_lines(df):
    fig = px.scatter(
        df,
        x='Pct_Buried_Lines',
        y='mean_abs_residual',
        title='Mean Absolute Residual vs. Percent of Buried Power Lines',
        labels={'Pct_Buried_Lines': 'Percent of Buried Power Lines', 'mean_abs_residual': 'Mean Absolute Residual'},
        template='plotly_white'
    )

    fig.update_layout(height=800)

    # Show the plot
    fig.show()
```
