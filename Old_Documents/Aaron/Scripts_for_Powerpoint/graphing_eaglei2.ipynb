```python
# Load the eaglei data
outages = pd.read_parquet('../Data/eaglei_data/eaglei_outages.parquet')
```

```python
#Load the US Census Counties shapefile
counties = geopandas.read_file('../Data/County_Shapefiles/cb_2023_us_county_500k')
```

```python
# Normalize colormap to the data range
max_outages = outages['customers_out'].max()
min_outages = outages['customers_out'].min()
```

```python
# Merge outages on fips_code and counties on GEOID as an integer
counties['GEOID'] = counties['GEOID'].astype(int)
merged = outages.merge(counties, left_on='fips_code', right_on='GEOID', how='left')
```

```python
# The datetime column lists the year-month-day. restrict merged to the entries between 2018-01-01 and 2018-12-31
merged_2018 = merged[(merged['datetime'] >= '2018-01-01') & (merged['datetime'] <= '2018-12-31')]

```

```python
fig = px.choropleth(merged_2018, locations='geometry', color='customers_out', animation_frame = 'datetime')
```
